# Task
Analyze the water quality data from "/4433201-water_potability.pdf" using pandas, numpy, and linear regression to perform a comprehensive multi-parameter assessment.

**Reasoning**:
Load the data from the CSV file, convert columns to numeric, display the head, info, and null values, fill missing values with the mean, and display null values again to verify. This addresses all the instructions for this subtask.



```python
import pandas as pd

# Load the data from the CSV file, specifying no header and providing column names
file_path = '/4433201-water_potability.csv'
column_names = ['ph', 'Hardness', 'Solids', 'Chloramines', 'Sulfate', 'Conductivity', 'Organic_carbon', 'Trihalomethanes', 'Turbidity', 'Potability']
df = pd.read_csv(file_path, sep=',', header=None, names=column_names)

# Convert relevant columns to numeric, coercing errors
for col in column_names:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Display the first 5 rows
display(df.head())

# Display the data types of each column
display(df.info())

# Check for missing values
display(df.isnull().sum())

# Handle missing values by filling them with the mean
df.fillna(df.mean(), inplace=True)

# Verify that missing values have been filled
display(df.isnull().sum())
```

## Explore data

### Subtask:
Perform exploratory data analysis (EDA) to understand the distribution of parameters and their relationships using visualizations and descriptive statistics.


**Reasoning**:
The previous steps successfully loaded the data and handled missing values. Now, I need to perform exploratory data analysis as per the subtask instructions. This involves displaying descriptive statistics, generating histograms, creating pairwise scatter plots, and calculating the correlation matrix.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Display descriptive statistics
display(df.describe())

# 2. Generate histograms for numerical columns
df.hist(figsize=(15, 10))
plt.tight_layout()
plt.show()

# 3. Create pairwise scatter plots for a subset of columns
# Check if 'Potability' column has any non-null values before using it as hue
if df['Potability'].isnull().all():
    subset_cols = ['ph', 'Hardness', 'Solids', 'Chloramines']
    sns.pairplot(df[subset_cols])
    plt.show()
else:
    subset_cols = ['ph', 'Hardness', 'Solids', 'Chloramines', 'Potability']
    sns.pairplot(df[subset_cols], hue='Potability')
    plt.show()


# 4. Calculate and display the correlation matrix
correlation_matrix = df.corr()
display(correlation_matrix)
```

## Cluster water samples (optional)

### Subtask:
If appropriate, consider clustering water samples based on their parameter values to identify different water quality profiles.


**Reasoning**:
Import the necessary library for clustering, select features, determine the number of clusters, instantiate and fit the KMeans model, add cluster labels to the DataFrame, and display the count and mean of samples in each cluster.



```python
from sklearn.cluster import KMeans
import numpy as np

# Select the features for clustering (all columns except 'Potability')
# Ensure X contains only numeric columns and handle potential NaNs introduced during previous steps if any
X = df.drop('Potability', axis=1)

# Drop columns with all NaN values or handle them appropriately for clustering
X.dropna(axis=1, how='all', inplace=True)

# Ensure remaining data has no NaNs - fill with mean if any were not handled earlier
X.fillna(X.mean(), inplace=True)


# Determine an appropriate number of clusters (starting with 3)
n_clusters = 3

# Instantiate a KMeans object
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10) # Set n_init explicitly

# Fit the KMeans model to the selected features
kmeans.fit(X)

# Add the cluster labels to the DataFrame
df['Cluster'] = kmeans.labels_

# Display the count of samples in each cluster
print("Count of samples in each cluster:")
display(df['Cluster'].value_counts())

# Calculate and display the mean values of each feature for each cluster
print("\nMean values of each feature for each cluster:")
display(df.groupby('Cluster').mean())
```
